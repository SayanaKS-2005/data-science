# **Array properties**

*ndim- number of dimensions-arr.ndim

*shape-size of each dimensions-arr.shape

*size-total number of elements-arr.size

*dtypw-data type of elements-arr.dtype

```python
print("Number of dimensions ",arr.ndim)
print("size of each dimension ",arr.shape)
print("total number of elements ",arr.size)
print("data type of elements ",arr.dtype)
```

# **linspace()**
it creeates number that are equally spaced between a start and an end value.

np.linspace(start,stop,num).

*start-starting value

*stop-ending value

*num-how many values you want

```python
np.linspace(0,10,5)
```

```python
np.linspace(0, 10, 5, endpoint=True)
```

```python
a = np.linspace(0, 10, 5, endpoint=False)
print(a)
```
